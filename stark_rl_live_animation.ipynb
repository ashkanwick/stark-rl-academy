## 🤖 بخش ۱ — پایدارساز پرواز (Q-Learning تک‌عامله)
صحنه‌ی «اولین پرواز»: محور کج‌شدگی از −۴ تا +۴ — دیوارها 💥 یعنی سقوط، نقطه‌ی 0 یعنی تعادل کامل.
آموزش ۸۰۰ اپیزودی فقط ~۱ ثانیه طول می‌کشد:

```python
env1 = StarkFlightBalancerEnv()
print("🧠 TRAINING J.A.R.V.I.S — 800 episodes of trial & error ...")
Q1, rewards1, crashes1 = train_flight_stabilizer(env1, episodes=800)
print_learned_policy(env1, Q1)
```

### 🎬 انیمیشن زنده — HUD جارویس (فریم‌به‌فریم)
زره از شیب −۳ شروع می‌کند؛ تلاطم باد 🌪 چند بار ناگهانی آن را هل می‌دهد و جارویس هر بار با رانشگرهای 🔥 شیب را اصلاح می‌کند. مقادیر `Q` و سیاست آموخته‌شده هم زنده نمایش داده می‌شوند. **این سلول را هر بار دوباره اجرا کنید!**

```python
animate_flight_stabilizer(Q1, env1)   # 15 گام، هر فریم 0.6 ثانیه
```

## 🚀 بخش ۲ — پروتکل مهمانی خانگی (Q-Learning چندعامله)
شبکه‌ی ۶×۶ با ۳ زره‌ی مستقل که فقط رادار هم‌تیمی‌هایشان را می‌بینند — آموزش ۱۵۰۰ اپیزود:

```python
env2 = HousePartyGridEnv()
print("🧠 TRAINING 3 INDEPENDENT Q-TABLES — 1500 episodes ...")
Qs2, team2, coll2, succ2 = train_house_party(env2, episodes=1500)
```

### 🎬 انیمیشن زنده — رادار چندعامله (یک فریم در هر راند)
هر ۳ زره هم‌زمان (نوبتی) با سیاست آموخته‌شده‌ی **خودشان** حرکت می‌کنند، از آتش 💥 رد می‌شوند، با هم برخورد نمی‌کنند و روی رآکتور 🔋 خودشان فرود می‌آیند.

```python
animate_house_party(Qs2, env2)   # یک فریم در هر راند، 0.5 ثانیه
```
